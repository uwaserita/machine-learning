```python
import numpy as np
import pandas as pd
```

<h1>Data Procesing</h1>

<h2>1. Read the loan-train.csv file</h2>

```python
# Write your code here
df = pd.read_csv("loan-train.csv")
df
```

<h2>2. Drop the Loan_ID column</h2>

```python
# Write your code here
df = df.drop('Loan_ID', axis=1)
df
```

<h2>3. Preprocess the dataset (Missing values, Incorrect formnat, ....)</h2>

<h3>3.0 Handle the Gender inconsistancies</h3>

```python
# replace M with Male and F with Female
df['Gender'] = df['Gender'].replace({'M': 'Male', 'F':'Female'})
df['Gender'].unique()
```

<h3>3.1. identifying missing Values in each column</h3>

```python
# Write your code here
df.isnull().sum()
```

<h3>3.2. Filling all the missing Values</h3>

```python
# Write your code here
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Married'] = df['Married'].fillna(df['Married'].mode()[0])
df['Dependents'] = df['Dependents'].fillna(df['Dependents'].mode()[0])
df['Self_Employed'] = df['Self_Employed'].fillna(df['Self_Employed'].mode()[0])
df['LoanAmount'] = df['LoanAmount'].fillna(df['LoanAmount'].median())
df['Loan_Amount_Term'] = df['Loan_Amount_Term'].fillna(df['Loan_Amount_Term'].mode()[0])
df['Credit_History'] = df['Credit_History'].fillna(df['Credit_History'].mode()[0])
```

<h3> 3.3. Check all missing vallues have been filles</h3>

```python
# Write your code here
df.isnull().sum()
```

<h2>4. Separate the independant & Depedant variables</h3>

```python
x = df.iloc[: , :-1]
y = df.iloc[: , -1]
```

```python
print(x.shape)
print(y.shape)
```

<h2>4. Encode all categorical Features</h2>

<h3>4.1. Generate Dummy features for all categorical features</h3>

```python
#Creating Dummy variables
dum_gender = pd.get_dummies(x['Gender'], dtype = int)
dum_Married = pd.get_dummies(x['Married'], dtype = int)
dum_Dependents = pd.get_dummies(x['Dependents'], dtype = int)
dum_Education = pd.get_dummies(x['Education'], dtype = int)
dum_Self_Employed = pd.get_dummies(x['Self_Employed'], dtype = int)
dum_Property_Area = pd.get_dummies(x['Property_Area'], dtype = int)
```

```python
dum_Education
```

<h3>4.2. Rename the indicator (Dummy) variables for clarity</h3>

```python
# change dumy columns names for clarity
dum_Married = dum_Married.rename(columns={'Yes': 'Maried', 'No':'Single'})
dum_Self_Employed = dum_Self_Employed.rename(columns={'Yes': 'Self_Empl', 'No':'Employed'})
```

<h3>4.3. Attache the dummy features to the original dataframe</h3>

```python
#attach the dumy features to the orogonal set
x = pd.concat([x, dum_gender, dum_Married, dum_Dependents, dum_Education, dum_Self_Employed, dum_Property_Area], axis=1)
x.head()
```

<h3>4.4. Drop the oroginal categorical features and see the result</h3>

```python
# Drop the oroginal categorical features and display the resulting df

x = x.drop(['Gender', 'Married', 'Dependents', 'Education', 'Self_Employed', 'Property_Area'], axis=1)
x
```

<h2>6. Scale the features </h2>

```python
# Feature scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x = sc.fit_transform(x)
x
```

<h2>7. Split the data set into Trainn & Test Sets (80-20)</h2>

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
print(y_train.shape)
print(x.shape)
```

<h2>8. Build model using train dataset by applying the Logistic Regressio</h2>

```python
from sklearn.linear_model import LogisticRegression
lg = LogisticRegression(random_state = 20)
lg.fit(x_train, y_train)
```

<h2>8. Test the model using test dataset</h2>

```python
lg.fit(x_train, y_train)
y_pred = lg.predict (x_test)
y_comp = pd.DataFrame({'y_pred': y_pred,'y_test': y_test})
y_comp
```

```python
y_pred
```

<h2>Generate the COnfusion Matrix</h2>

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```
